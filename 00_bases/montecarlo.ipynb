```python
import math
# in origin
r = 3
real_circle_area = (math.pi * (r ** 2))
print(real_circle_area)
```

```python
import random

def is_inside_circle(x, y, r):
    return ((x ** 2) + (y ** 2)) <= (r ** 2)


container_area = ((2 * r) ** 2)

points_to_generate = 1000000
hit_points = 0

for _ in range(points_to_generate):
    x = -3 + (random.random() * 6)
    y = -3 + (random.random() * 6)

    if is_inside_circle(x, y, r):
        hit_points += 1

print(f"Estimated area: {(hit_points / points_to_generate) * container_area}")
print(f"Real área: {real_circle_area}")
```
